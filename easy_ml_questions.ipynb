### 1. What is the difference between a list and a NumPy array in Python? Demonstrate with an example.

```python
# Your code/notes here
import numpy as np
list0 = [1, 2.5, 'monke', 'p', 42069]
print(f"Lists can have different data types: {list0}")

array = np.array([1.00,0.72,3.14,2.25,0.00])
print(f"Arrays can only have one data type: {array} of type {type(array)}")
print("Array dtype:", array.dtype)

import time

size = 1000000000
list_data = list(range(size))
t1 = time.time()
list_sqr = [x**2 for x in list_data]
t2 = time.time()
print(f"Time taken for list: {t2-t1} seconds")

array_data = np.array(list_data)
t3 = time.time()
array_sqr = array_data**2 #vectorized operation apparently
t4 = time.time()
print(f"Time taken for array: {t4-t3} seconds")

#lists require loops for operations
#arrays are faster because they are vectorized
l1 = [1,2,3,4,5]
l2 = [6,7,8,9,10]
t5 = time.time()
l3 = [x+y for x,y in zip(l1,l2)]
t6 = time.time()
print(f"List operation: {l3}")
print(f"Time taken for list: {t6-t5} seconds")

a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
t7 = time.time()
a3 = a1+a2
t8 = time.time()
print(f"Array operation: {a3}")
print(f"Time taken for array: {t8-t7} seconds")

#arrays are faster because they are vectorized
#lists require loops for operations
#arrays are faster because they are vectorized
```

### 6. Explain the difference between `loc` and `iloc` in pandas with examples.

```python
# Your code/notes here
import pandas as pd

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 32, 18]
}, index=['a', 'b', 'c'])

# Get row with label 'b'
print(df.loc['b'])

# Get multiple rows by label
print(df.loc[['a', 'c']])

# Get rows where age > 20
print(df.loc[df['age'] > 20])


########################################
# Get row at index position 1 (i.e., second row)
print(df.iloc[1])

# Get first two rows
print(df.iloc[0:2])

# Get first row and first column value
print(df.iloc[0, 0])  # Alice

```
